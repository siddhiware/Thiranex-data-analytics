```python

# ============================================================
# THIRANEX INTERNSHIP - TASK 4
# DATA CLEANING & REPORTING AUTOMATION
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import zipfile

np.random.seed(42)

print("Libraries imported successfully!")


# ============================================================
# 2. CREATE A RAW / MESSY DATASET
# ============================================================

n = 350

names = [
    "Aarav Sharma",
    "Priya Patil",
    "Rahul Deshmukh",
    "Sneha Kulkarni",
    "Amit Joshi",
    "Neha Pawar",
    "Rohan Jadhav",
    "Pooja More",
    "Karan Shah",
    "Anjali Gupta",
    "Vikas Kale",
    "Meera Nair"
]

cities = [
    "Pune",
    "pune",
    "Mumbai",
    "mumbai",
    "Delhi",
    "Bangalore",
    "Hyderabad",
    None
]

departments = [
    "Sales",
    "sales",
    "Marketing",
    "marketing",
    "IT",
    "Finance",
    None
]

products = [
    "Laptop",
    "Laptop ",
    "Smartphone",
    "smartphone",
    "Tablet",
    "Headphones",
    "Monitor",
    None
]

payments = [
    "UPI",
    "upi",
    "Credit Card",
    "Credit card",
    "Cash",
    "Debit Card",
    None
]

dates = pd.date_range(
    start="2025-01-01",
    end="2026-08-31",
    freq="D"
)

raw_data = pd.DataFrame({

    "Record_ID": np.arange(1, n + 1),

    "Customer_Name": np.random.choice(
        names, n
    ),

    "Order_Date": np.random.choice(
        dates, n
    ),

    "City": np.random.choice(
        cities, n
    ),

    "Department": np.random.choice(
        departments, n
    ),

    "Product": np.random.choice(
        products, n
    ),

    "Quantity": np.random.choice(
        [1, 2, 3, 4, 5, np.nan],
        n
    ),

    "Unit_Price": np.random.choice(
        [499, 999, 1499, 2499, 4999, 9999, np.nan],
        n
    ),

    "Payment_Method": np.random.choice(
        payments, n
    )
})


# Calculate sales
raw_data["Sales"] = (
    raw_data["Quantity"] *
    raw_data["Unit_Price"]
)


# Convert date to text
raw_data["Order_Date"] = (
    raw_data["Order_Date"]
    .dt.strftime("%Y-%m-%d")
)


# ------------------------------------------------------------
# Add intentional inconsistencies
# ------------------------------------------------------------

raw_data.loc[5, "City"] = " PUNE "
raw_data.loc[12, "Department"] = " SALES "
raw_data.loc[18, "Payment_Method"] = " UPI "
raw_data.loc[25, "Product"] = " smartphone"


# ------------------------------------------------------------
# Add duplicate records
# ------------------------------------------------------------

duplicate_rows = raw_data.iloc[
    [10, 20, 30, 40, 50]
]

raw_data = pd.concat(
    [raw_data, duplicate_rows],
    ignore_index=True
)


# ------------------------------------------------------------
# Save RAW CSV
# ------------------------------------------------------------

raw_data.to_csv(
    "thiranex_task4_raw_messy_data.csv",
    index=False
)

print("\nRaw dataset created!")
print("Rows:", len(raw_data))

display(raw_data.head(10))


# ============================================================
# 3. INSPECT DATA
# ============================================================

print("\nDATA INFORMATION")
print("=" * 50)

print(raw_data.info())


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES")
print("=" * 50)

missing_values = raw_data.isnull().sum()

print(missing_values)


# ============================================================
# 5. CHECK DUPLICATES
# ============================================================

duplicate_count = raw_data.duplicated().sum()

print("\nDUPLICATE ROWS")
print("=" * 50)

print(
    "Number of duplicate rows:",
    duplicate_count
)


# ============================================================
# 6. DATA CLEANING
# ============================================================

clean_data = raw_data.copy()


# ------------------------------------------------------------
# Standardize text columns
# ------------------------------------------------------------

text_columns = [
    "Customer_Name",
    "City",
    "Department",
    "Product",
    "Payment_Method"
]

for column in text_columns:

    clean_data[column] = (
        clean_data[column]
        .astype("string")
        .str.strip()
        .str.title()
    )


# ------------------------------------------------------------
# Standardize payment methods
# ------------------------------------------------------------

clean_data["Payment_Method"] = (
    clean_data["Payment_Method"]
    .replace({
        "Upi": "UPI",
        "Credit Card": "Credit Card",
        "Debit Card": "Debit Card"
    })
)


# ============================================================
# 7. CONVERT DATA TYPES
# ============================================================

clean_data["Order_Date"] = pd.to_datetime(
    clean_data["Order_Date"],
    errors="coerce"
)

clean_data["Quantity"] = pd.to_numeric(
    clean_data["Quantity"],
    errors="coerce"
)

clean_data["Unit_Price"] = pd.to_numeric(
    clean_data["Unit_Price"],
    errors="coerce"
)


# ============================================================
# 8. HANDLE MISSING VALUES
# ============================================================

# Categorical columns

categorical_columns = [
    "City",
    "Department",
    "Product",
    "Payment_Method"
]

for column in categorical_columns:

    clean_data[column] = (
        clean_data[column]
        .fillna("Unknown")
    )


# Customer name

clean_data["Customer_Name"] = (
    clean_data["Customer_Name"]
    .fillna("Unknown Customer")
)


# Numerical columns

clean_data["Quantity"] = (
    clean_data["Quantity"]
    .fillna(
        clean_data["Quantity"].median()
    )
)

clean_data["Unit_Price"] = (
    clean_data["Unit_Price"]
    .fillna(
        clean_data["Unit_Price"].median()
    )
)


# Date

clean_data["Order_Date"] = (
    clean_data["Order_Date"]
    .fillna(
        clean_data["Order_Date"].median()
    )
)


# ============================================================
# 9. RECALCULATE SALES
# ============================================================

clean_data["Sales"] = (
    clean_data["Quantity"] *
    clean_data["Unit_Price"]
)


# ============================================================
# 10. REMOVE DUPLICATES
# ============================================================

before_duplicates = len(clean_data)

clean_data = (
    clean_data
    .drop_duplicates()
    .reset_index(drop=True)
)

after_duplicates = len(clean_data)

duplicates_removed = (
    before_duplicates -
    after_duplicates
)


print("\nDuplicates removed:",
      duplicates_removed)


# ============================================================
# 11. ADD REPORTING COLUMNS
# ============================================================

clean_data["Month"] = (
    clean_data["Order_Date"]
    .dt.to_period("M")
    .astype(str)
)

clean_data["Year"] = (
    clean_data["Order_Date"]
    .dt.year
)

clean_data["Quarter"] = (
    "Q" +
    clean_data["Order_Date"]
    .dt.quarter
    .astype(str)
)


# ============================================================
# 12. SAVE CLEANED DATASET
# ============================================================

clean_data.to_csv(
    "thiranex_task4_cleaned_data.csv",
    index=False
)

print("\nCleaned dataset saved!")

display(clean_data.head(10))


# ============================================================
# 13. DATA QUALITY REPORT
# ============================================================

quality_report = pd.DataFrame({

    "Column":
        clean_data.columns,

    "Data_Type":
        [
            str(clean_data[column].dtype)
            for column in clean_data.columns
        ],

    "Missing_Values":
        [
            clean_data[column].isnull().sum()
            for column in clean_data.columns
        ],

    "Unique_Values":
        [
            clean_data[column].nunique()
            for column in clean_data.columns
        ]
})


quality_report.to_csv(
    "thiranex_task4_data_quality_report.csv",
    index=False
)

print("\nDATA QUALITY REPORT")

display(quality_report)


# ============================================================
# 14. AUTOMATED SUMMARY REPORT
# ============================================================

initial_rows = len(raw_data)

initial_missing = (
    raw_data.isnull()
    .sum()
    .sum()
)

final_rows = len(clean_data)

remaining_missing = (
    clean_data.isnull()
    .sum()
    .sum()
)

total_sales = (
    clean_data["Sales"]
    .sum()
)

average_order_value = (
    clean_data["Sales"]
    .mean()
)

total_quantity = (
    clean_data["Quantity"]
    .sum()
)

unique_customers = (
    clean_data["Customer_Name"]
    .nunique()
)

unique_products = (
    clean_data["Product"]
    .nunique()
)

unique_cities = (
    clean_data["City"]
    .nunique()
)


summary_report = pd.DataFrame({

    "Metric": [

        "Initial Rows",

        "Initial Missing Values",

        "Duplicate Rows Removed",

        "Final Clean Rows",

        "Missing Values After Cleaning",

        "Total Sales",

        "Average Order Value",

        "Total Quantity Sold",

        "Unique Customers",

        "Unique Products",

        "Unique Cities"

    ],

    "Value": [

        initial_rows,

        initial_missing,

        duplicates_removed,

        final_rows,

        remaining_missing,

        round(total_sales, 2),

        round(average_order_value, 2),

        round(total_quantity, 2),

        unique_customers,

        unique_products,

        unique_cities

    ]
})


summary_report.to_csv(
    "thiranex_task4_summary_report.csv",
    index=False
)


print("\nSUMMARY REPORT")

display(summary_report)


# ============================================================
# 15. MONTHLY REPORT
# ============================================================

monthly_report = (

    clean_data
    .groupby("Month")
    .agg(

        Total_Orders=(
            "Record_ID",
            "count"
        ),

        Total_Quantity=(
            "Quantity",
            "sum"
        ),

        Total_Sales=(
            "Sales",
            "sum"
        ),

        Average_Order_Value=(
            "Sales",
            "mean"
        )

    )
    .reset_index()
)


monthly_report["Total_Sales"] = (
    monthly_report["Total_Sales"]
    .round(2)
)

monthly_report["Average_Order_Value"] = (
    monthly_report["Average_Order_Value"]
    .round(2)
)


monthly_report.to_csv(
    "thiranex_task4_monthly_report.csv",
    index=False
)


print("\nMONTHLY REPORT")

display(monthly_report)


# ============================================================
# 16. PRODUCT REPORT
# ============================================================

product_report = (

    clean_data
    .groupby("Product")
    .agg(

        Orders=(
            "Record_ID",
            "count"
        ),

        Quantity_Sold=(
            "Quantity",
            "sum"
        ),

        Total_Sales=(
            "Sales",
            "sum"
        ),

        Average_Sales=(
            "Sales",
            "mean"
        )

    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)


product_report["Total_Sales"] = (
    product_report["Total_Sales"]
    .round(2)
)

product_report["Average_Sales"] = (
    product_report["Average_Sales"]
    .round(2)
)


product_report.to_csv(
    "thiranex_task4_product_report.csv",
    index=False
)


print("\nPRODUCT REPORT")

display(product_report)


# ============================================================
# 17. CITY REPORT
# ============================================================

city_report = (

    clean_data
    .groupby("City")
    .agg(

        Orders=(
            "Record_ID",
            "count"
        ),

        Total_Sales=(
            "Sales",
            "sum"
        )

    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)


city_report["Total_Sales"] = (
    city_report["Total_Sales"]
    .round(2)
)


city_report.to_csv(
    "thiranex_task4_city_report.csv",
    index=False
)


print("\nCITY REPORT")

display(city_report)


# ============================================================
# 18. VISUALIZATION 1 — MISSING VALUES
# ============================================================

plt.figure(figsize=(10, 5))

raw_missing = (
    raw_data.isnull()
    .sum()
)

raw_missing = raw_missing[
    raw_missing > 0
]

raw_missing.plot(
    kind="bar"
)

plt.title(
    "Missing Values Before Data Cleaning"
)

plt.xlabel("Column")

plt.ylabel(
    "Number of Missing Values"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 19. VISUALIZATION 2 — MONTHLY SALES
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_report["Month"],
    monthly_report["Total_Sales"],
    marker="o"
)

plt.title(
    "Monthly Sales After Data Cleaning"
)

plt.xlabel("Month")

plt.ylabel("Sales")

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 20. VISUALIZATION 3 — TOP PRODUCTS
# ============================================================

plt.figure(figsize=(10, 6))

top_products = (
    product_report
    .head(7)
)

plt.bar(
    top_products["Product"],
    top_products["Total_Sales"]
)

plt.title(
    "Sales by Product"
)

plt.xlabel("Product")

plt.ylabel("Total Sales")

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. VISUALIZATION 4 — SALES BY CITY
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    city_report["City"],
    city_report["Total_Sales"]
)

plt.title(
    "Sales by City"
)

plt.xlabel("City")

plt.ylabel("Total Sales")

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 22. AUTOMATED BUSINESS INSIGHTS
# ============================================================

best_product = (
    product_report
    .iloc[0]["Product"]
)

best_product_sales = (
    product_report
    .iloc[0]["Total_Sales"]
)

best_city = (
    city_report
    .iloc[0]["City"]
)

best_city_sales = (
    city_report
    .iloc[0]["Total_Sales"]
)

best_month = (
    monthly_report
    .loc[
        monthly_report["Total_Sales"].idxmax(),
        "Month"
    ]
)


print("\n")
print("=" * 60)
print("AUTOMATED BUSINESS INSIGHTS")
print("=" * 60)

print(
    f"1. Highest-selling product: "
    f"{best_product}"
)

print(
    f"   Product sales: ₹{best_product_sales:,.2f}"
)

print(
    f"\n2. Highest-sales city: "
    f"{best_city}"
)

print(
    f"   City sales: ₹{best_city_sales:,.2f}"
)

print(
    f"\n3. Highest-sales month: "
    f"{best_month}"
)

print(
    f"\n4. Total cleaned sales: "
    f"₹{total_sales:,.2f}"
)

print(
    f"\n5. Duplicate records removed: "
    f"{duplicates_removed}"
)

print(
    f"\n6. Missing values remaining: "
    f"{remaining_missing}"
)


# ============================================================
# 23. CREATE ZIP FILE CONTAINING ALL CSV FILES
# ============================================================

csv_files = [

    "thiranex_task4_raw_messy_data.csv",

    "thiranex_task4_cleaned_data.csv",

    "thiranex_task4_data_quality_report.csv",

    "thiranex_task4_summary_report.csv",

    "thiranex_task4_monthly_report.csv",

    "thiranex_task4_product_report.csv",

    "thiranex_task4_city_report.csv"

]


with zipfile.ZipFile(
    "thiranex_task4_all_csv_files.zip",
    "w"
) as zip_file:

    for file in csv_files:

        zip_file.write(file)


# ============================================================
# 24. DISPLAY ALL CREATED FILES
# ============================================================

print("\n")
print("=" * 60)
print("TASK 4 COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nCreated files:")

for file in csv_files:
    print("✓", file)

print(
    "✓ thiranex_task4_all_csv_files.zip"
)

print("\nYou can now download these files from Colab.")
```
