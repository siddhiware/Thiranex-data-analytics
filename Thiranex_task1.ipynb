```python

# ============================================================
# THIRANEX INTERNSHIP PROJECT
# SALES & REVENUE ANALYSIS DASHBOARD
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from google.colab import files

# For reproducible results
np.random.seed(42)


# -----------------------------
# 2. CREATE SALES DATASET
# -----------------------------

n = 500

products = [
    "Laptop",
    "Smartphone",
    "Tablet",
    "Monitor",
    "Keyboard",
    "Mouse",
    "Headphones",
    "Printer",
    "Webcam",
    "USB Drive"
]

categories = {
    "Laptop": "Electronics",
    "Smartphone": "Electronics",
    "Tablet": "Electronics",
    "Monitor": "Electronics",
    "Keyboard": "Accessories",
    "Mouse": "Accessories",
    "Headphones": "Accessories",
    "Printer": "Office Equipment",
    "Webcam": "Accessories",
    "USB Drive": "Storage"
}

prices = {
    "Laptop": 55000,
    "Smartphone": 30000,
    "Tablet": 22000,
    "Monitor": 15000,
    "Keyboard": 1800,
    "Mouse": 900,
    "Headphones": 2500,
    "Printer": 12000,
    "Webcam": 3500,
    "USB Drive": 1200
}

regions = ["North", "South", "East", "West"]
customers = ["Consumer", "Small Business", "Enterprise"]

dates = pd.date_range(
    start="2025-01-01",
    end="2026-08-31",
    periods=n
)

data = []

for i in range(n):

    product = np.random.choice(products)
    quantity = np.random.randint(1, 8)
    unit_price = prices[product]

    # Add small price variation
    unit_price = round(
        unit_price * np.random.uniform(0.90, 1.10),
        2
    )

    sales = round(quantity * unit_price, 2)

    data.append([
        f"ORD-{1001+i}",
        dates[i],
        product,
        categories[product],
        np.random.choice(regions),
        np.random.choice(customers),
        quantity,
        unit_price,
        sales
    ])


columns = [
    "Order_ID",
    "Order_Date",
    "Product",
    "Category",
    "Region",
    "Customer_Segment",
    "Quantity",
    "Unit_Price",
    "Sales"
]

df = pd.DataFrame(data, columns=columns)


# -----------------------------
# 3. ADD REVENUE COLUMN
# -----------------------------

df["Revenue"] = df["Sales"]


# -----------------------------
# 4. SAVE CSV FILE
# -----------------------------

csv_file = "thiranex_sales_revenue_data.csv"

df.to_csv(csv_file, index=False)

print("CSV file created successfully!")
print("File name:", csv_file)


# -----------------------------
# 5. DISPLAY DATA
# -----------------------------

print("\nFirst 10 records:")
display(df.head(10))

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Information:")
print(df.info())


# -----------------------------
# 6. DATA CLEANING
# -----------------------------

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# Remove duplicates if any
df = df.drop_duplicates()

# Convert date
df["Order_Date"] = pd.to_datetime(df["Order_Date"])


# -----------------------------
# 7. KPI CALCULATIONS
# -----------------------------

total_sales = df["Sales"].sum()
total_revenue = df["Revenue"].sum()
total_orders = df["Order_ID"].nunique()
total_quantity = df["Quantity"].sum()
average_order_value = total_sales / total_orders

print("\n==============================")
print("SALES & REVENUE KPIs")
print("==============================")

print(f"Total Sales       : ₹{total_sales:,.2f}")
print(f"Total Revenue     : ₹{total_revenue:,.2f}")
print(f"Total Orders      : {total_orders:,}")
print(f"Total Quantity    : {total_quantity:,}")
print(f"Average Order Value: ₹{average_order_value:,.2f}")


# -----------------------------
# 8. MONTHLY SALES ANALYSIS
# -----------------------------

df["Month"] = df["Order_Date"].dt.to_period("M")

monthly_sales = (
    df.groupby("Month")["Sales"]
    .sum()
    .reset_index()
)

monthly_sales["Month"] = monthly_sales["Month"].astype(str)

print("\nMonthly Sales:")
display(monthly_sales)


# -----------------------------
# 9. PRODUCT ANALYSIS
# -----------------------------

product_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Product:")
display(product_sales)


# -----------------------------
# 10. TOP 10 PRODUCTS
# -----------------------------

top_products = product_sales.head(10)

print("\nTop Performing Products:")
display(top_products)


# -----------------------------
# 11. CATEGORY ANALYSIS
# -----------------------------

category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Category:")
display(category_sales)


# -----------------------------
# 12. REGION ANALYSIS
# -----------------------------

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Region:")
display(region_sales)


# -----------------------------
# 13. CUSTOMER SEGMENT ANALYSIS
# -----------------------------

customer_sales = (
    df.groupby("Customer_Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Customer Segment:")
display(customer_sales)


# ============================================================
# VISUALIZATIONS
# ============================================================


# -----------------------------
# 14. MONTHLY REVENUE TREND
# -----------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales / Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()


# -----------------------------
# 15. TOP PRODUCTS
# -----------------------------

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(
    kind="barh"
)

plt.title("Top Performing Products")
plt.xlabel("Sales (₹)")
plt.ylabel("Product")

plt.tight_layout()
plt.show()


# -----------------------------
# 16. CATEGORY SALES
# -----------------------------

plt.figure(figsize=(9, 5))

category_sales.plot(
    kind="bar"
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# -----------------------------
# 17. REGION SALES
# -----------------------------

plt.figure(figsize=(8, 5))

region_sales.plot(
    kind="bar"
)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales (₹)")

plt.tight_layout()
plt.show()


# -----------------------------
# 18. CUSTOMER SEGMENT
# -----------------------------

plt.figure(figsize=(8, 5))

customer_sales.plot(
    kind="bar"
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Sales (₹)")

plt.tight_layout()
plt.show()


# -----------------------------
# 19. CATEGORY PIE CHART
# -----------------------------

plt.figure(figsize=(7, 7))

category_sales.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Revenue Distribution by Category")
plt.ylabel("")

plt.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

best_product = product_sales.idxmax()
best_product_sales = product_sales.max()

best_category = category_sales.idxmax()
best_category_sales = category_sales.max()

best_region = region_sales.idxmax()
best_region_sales = region_sales.max()

best_customer = customer_sales.idxmax()
best_customer_sales = customer_sales.max()


print("\n==========================================")
print("BUSINESS INSIGHTS")
print("==========================================")

print(
    f"1. The highest-selling product is {best_product} "
    f"with sales of ₹{best_product_sales:,.2f}."
)

print(
    f"2. The best-performing category is {best_category} "
    f"with sales of ₹{best_category_sales:,.2f}."
)

print(
    f"3. The highest-performing region is {best_region} "
    f"with sales of ₹{best_region_sales:,.2f}."
)

print(
    f"4. The largest customer segment is {best_customer} "
    f"with sales of ₹{best_customer_sales:,.2f}."
)

print(
    f"5. The average order value is ₹{average_order_value:,.2f}."
)


# ============================================================
# 20. CREATE KPI SUMMARY CSV
# ============================================================

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Revenue",
        "Total Orders",
        "Total Quantity",
        "Average Order Value"
    ],
    "Value": [
        total_sales,
        total_revenue,
        total_orders,
        total_quantity,
        average_order_value
    ]
})

kpi_summary.to_csv(
    "thiranex_kpi_summary.csv",
    index=False
)

print("\nKPI summary file created.")


# ============================================================
# 21. CREATE PRODUCT SUMMARY CSV
# ============================================================

product_summary = (
    df.groupby(["Product", "Category"])
    .agg(
        Total_Quantity=("Quantity", "sum"),
        Total_Sales=("Sales", "sum"),
        Average_Price=("Unit_Price", "mean")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

product_summary.to_csv(
    "thiranex_product_analysis.csv",
    index=False
)

print("Product analysis file created.")


# ============================================================
# 22. CREATE REGION SUMMARY CSV
# ============================================================

region_summary = (
    df.groupby("Region")
    .agg(
        Orders=("Order_ID", "nunique"),
        Quantity=("Quantity", "sum"),
        Sales=("Sales", "sum")
    )
    .reset_index()
    .sort_values("Sales", ascending=False)
)

region_summary.to_csv(
    "thiranex_region_analysis.csv",
    index=False
)

print("Region analysis file created.")


# ============================================================
# 23. DOWNLOAD MAIN CSV
# ============================================================

print("\nDownloading main dataset...")

files.download("thiranex_sales_revenue_data.csv")
```
