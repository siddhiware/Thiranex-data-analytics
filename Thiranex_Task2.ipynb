```python

# ============================================================
# THIRANEX INTERNSHIP - TASK 2
# CUSTOMER SEGMENTATION PROJECT
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from google.colab import files
from IPython.display import display

# Reproducible results
np.random.seed(42)

print("Libraries imported successfully!")


# ============================================================
# 2. CREATE CUSTOMER DATASET
# ============================================================

n_customers = 500

customer_ids = [
    f"CUST-{1001+i}"
    for i in range(n_customers)
]

# Demographic information
ages = np.random.randint(18, 66, n_customers)

genders = np.random.choice(
    ["Male", "Female"],
    n_customers
)

locations = np.random.choice(
    ["Pune", "Mumbai", "Delhi", "Bangalore",
     "Hyderabad", "Chennai", "Nagpur", "Nashik"],
    n_customers
)

# Income
annual_income = np.random.randint(
    200000,
    1800000,
    n_customers
)

# Customer purchase behavior
total_purchases = np.random.randint(
    1,
    50,
    n_customers
)

average_purchase_value = np.random.randint(
    300,
    25000,
    n_customers
)

total_spending = (
    total_purchases *
    average_purchase_value
)

# Recency:
# Number of days since last purchase
recency_days = np.random.randint(
    1,
    365,
    n_customers
)

# Customer tenure in months
tenure_months = np.random.randint(
    1,
    72,
    n_customers
)

# Preferred product category
categories = np.random.choice(
    [
        "Electronics",
        "Clothing",
        "Grocery",
        "Beauty",
        "Home & Kitchen",
        "Sports"
    ],
    n_customers
)

# Preferred payment method
payment_methods = np.random.choice(
    [
        "UPI",
        "Credit Card",
        "Debit Card",
        "Cash",
        "Net Banking"
    ],
    n_customers
)


# Create DataFrame
df = pd.DataFrame({

    "Customer_ID": customer_ids,

    "Age": ages,

    "Gender": genders,

    "Location": locations,

    "Annual_Income": annual_income,

    "Total_Purchases": total_purchases,

    "Average_Purchase_Value": average_purchase_value,

    "Total_Spending": total_spending,

    "Recency_Days": recency_days,

    "Tenure_Months": tenure_months,

    "Preferred_Category": categories,

    "Payment_Method": payment_methods

})


# ============================================================
# 3. SAVE ORIGINAL DATASET AS CSV
# ============================================================

original_file = "thiranex_customer_segmentation_data.csv"

df.to_csv(
    original_file,
    index=False
)

print("\nCustomer dataset created successfully!")

print("\nDataset shape:")
print(df.shape)

print("\nFirst 10 records:")

display(df.head(10))


# ============================================================
# 4. DATA INFORMATION
# ============================================================

print("\n==============================")
print("DATASET INFORMATION")
print("==============================")

print(df.info())


# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

print("\nMissing values:")

print(df.isnull().sum())


# ============================================================
# 6. CHECK DUPLICATES
# ============================================================

print("\nDuplicate records:")

print(df.duplicated().sum())


# Remove duplicates
df = df.drop_duplicates()


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

print("\n==============================")
print("DESCRIPTIVE STATISTICS")
print("==============================")

display(
    df.describe()
)


# ============================================================
# 8. RFM FEATURES
# ============================================================

# RFM:
#
# R = Recency
# F = Frequency
# M = Monetary

df["Recency"] = df["Recency_Days"]

df["Frequency"] = df["Total_Purchases"]

df["Monetary"] = df["Total_Spending"]


print("\nRFM features created!")

display(
    df[
        [
            "Customer_ID",
            "Recency",
            "Frequency",
            "Monetary"
        ]
    ].head()
)


# ============================================================
# 9. SELECT FEATURES FOR CLUSTERING
# ============================================================

features = [

    "Age",

    "Annual_Income",

    "Recency",

    "Frequency",

    "Monetary"

]

X = df[features].copy()


# ============================================================
# 10. STANDARDIZE DATA
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("\nData standardized successfully!")


# ============================================================
# 11. FIND OPTIMAL NUMBER OF CLUSTERS
# ============================================================

inertia = []

K = range(2, 11)

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(
        kmeans.inertia_
    )


# ============================================================
# 12. ELBOW METHOD GRAPH
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    K,
    inertia,
    marker="o"
)

plt.title(
    "Elbow Method for Optimal Number of Clusters"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.xticks(K)

plt.grid(True)

plt.show()


# ============================================================
# 13. SILHOUETTE SCORE
# ============================================================

silhouette_scores = {}

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores[k] = score


print("\n==============================")
print("SILHOUETTE SCORES")
print("==============================")

for k, score in silhouette_scores.items():

    print(
        f"K = {k}: "
        f"Silhouette Score = {score:.4f}"
    )


# ============================================================
# 14. SELECT NUMBER OF CLUSTERS
# ============================================================

# We use 4 customer segments for this project.
# You can change this after examining the Elbow graph
# and Silhouette scores.

optimal_k = 4


# ============================================================
# 15. APPLY K-MEANS CLUSTERING
# ============================================================

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    X_scaled
)


print("\nK-Means clustering completed!")


# ============================================================
# 16. CHECK CLUSTER COUNTS
# ============================================================

print("\nCustomer count in each cluster:")

cluster_counts = (
    df["Cluster"]
    .value_counts()
    .sort_index()
)

display(cluster_counts)


# ============================================================
# 17. CUSTOMER SEGMENT PROFILE
# ============================================================

segment_profile = (

    df.groupby("Cluster")

    .agg({

        "Age": "mean",

        "Annual_Income": "mean",

        "Recency": "mean",

        "Frequency": "mean",

        "Monetary": "mean",

        "Customer_ID": "count"

    })

    .round(2)

)


segment_profile = segment_profile.rename(
    columns={
        "Customer_ID": "Customer_Count"
    }
)


print("\n==============================")
print("CUSTOMER SEGMENT PROFILE")
print("==============================")

display(
    segment_profile
)


# ============================================================
# 18. AUTOMATICALLY NAME SEGMENTS
# ============================================================

# Determine characteristics of each cluster.

segment_names = {}

median_frequency = df["Frequency"].median()

median_monetary = df["Monetary"].median()

median_recency = df["Recency"].median()


for cluster in segment_profile.index:

    recency = segment_profile.loc[
        cluster,
        "Recency"
    ]

    frequency = segment_profile.loc[
        cluster,
        "Frequency"
    ]

    monetary = segment_profile.loc[
        cluster,
        "Monetary"
    ]


    if (
        frequency >= median_frequency
        and monetary >= median_monetary
        and recency <= median_recency
    ):

        segment_names[cluster] = (
            "High Value Customers"
        )


    elif (
        frequency >= median_frequency
        and monetary < median_monetary
    ):

        segment_names[cluster] = (
            "Frequent Customers"
        )


    elif (
        recency > median_recency
        and frequency < median_frequency
    ):

        segment_names[cluster] = (
            "At Risk Customers"
        )


    else:

        segment_names[cluster] = (
            "Regular Customers"
        )


# Add segment name
df["Customer_Segment"] = (
    df["Cluster"]
    .map(segment_names)
)


print("\nCustomer segment names:")

for cluster, name in segment_names.items():

    print(
        f"Cluster {cluster} → {name}"
    )


# ============================================================
# 19. FINAL SEGMENT SUMMARY
# ============================================================

final_segment_summary = (

    df.groupby("Customer_Segment")

    .agg(

        Customers=(
            "Customer_ID",
            "count"
        ),

        Average_Age=(
            "Age",
            "mean"
        ),

        Average_Income=(
            "Annual_Income",
            "mean"
        ),

        Average_Recency=(
            "Recency",
            "mean"
        ),

        Average_Frequency=(
            "Frequency",
            "mean"
        ),

        Average_Spending=(
            "Monetary",
            "mean"
        )

    )

    .round(2)

)


print("\n==============================")
print("FINAL CUSTOMER SEGMENT SUMMARY")
print("==============================")

display(
    final_segment_summary
)


# ============================================================
# 20. SEGMENT DISTRIBUTION
# ============================================================

segment_counts = (

    df["Customer_Segment"]

    .value_counts()

)


plt.figure(figsize=(9, 5))

segment_counts.plot(
    kind="bar"
)

plt.title(
    "Number of Customers in Each Segment"
)

plt.xlabel(
    "Customer Segment"
)

plt.ylabel(
    "Number of Customers"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. CUSTOMER SEGMENT PIE CHART
# ============================================================

plt.figure(figsize=(8, 8))

segment_counts.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title(
    "Customer Segment Distribution"
)

plt.ylabel("")

plt.show()


# ============================================================
# 22. SPENDING BY SEGMENT
# ============================================================

spending_by_segment = (

    df.groupby(
        "Customer_Segment"
    )["Monetary"]

    .mean()

    .sort_values(
        ascending=False
    )

)


plt.figure(figsize=(10, 5))

spending_by_segment.plot(
    kind="bar"
)

plt.title(
    "Average Customer Spending by Segment"
)

plt.xlabel(
    "Customer Segment"
)

plt.ylabel(
    "Average Spending (₹)"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 23. FREQUENCY BY SEGMENT
# ============================================================

frequency_by_segment = (

    df.groupby(
        "Customer_Segment"
    )["Frequency"]

    .mean()

    .sort_values(
        ascending=False
    )

)


plt.figure(figsize=(10, 5))

frequency_by_segment.plot(
    kind="bar"
)

plt.title(
    "Average Purchase Frequency by Segment"
)

plt.xlabel(
    "Customer Segment"
)

plt.ylabel(
    "Average Number of Purchases"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 24. RECENCY BY SEGMENT
# ============================================================

recency_by_segment = (

    df.groupby(
        "Customer_Segment"
    )["Recency"]

    .mean()

    .sort_values()

)


plt.figure(figsize=(10, 5))

recency_by_segment.plot(
    kind="bar"
)

plt.title(
    "Average Recency by Customer Segment"
)

plt.xlabel(
    "Customer Segment"
)

plt.ylabel(
    "Days Since Last Purchase"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 25. INCOME VS SPENDING
# ============================================================

plt.figure(figsize=(10, 6))

for segment in df[
    "Customer_Segment"
].unique():

    segment_data = df[
        df["Customer_Segment"] == segment
    ]

    plt.scatter(

        segment_data[
            "Annual_Income"
        ],

        segment_data[
            "Monetary"
        ],

        label=segment,

        alpha=0.7
    )


plt.title(
    "Annual Income vs Customer Spending"
)

plt.xlabel(
    "Annual Income (₹)"
)

plt.ylabel(
    "Total Spending (₹)"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 26. AGE VS SPENDING
# ============================================================

plt.figure(figsize=(10, 6))

for segment in df[
    "Customer_Segment"
].unique():

    segment_data = df[
        df["Customer_Segment"] == segment
    ]

    plt.scatter(

        segment_data["Age"],

        segment_data["Monetary"],

        label=segment,

        alpha=0.7

    )


plt.title(
    "Age vs Customer Spending"
)

plt.xlabel(
    "Age"
)

plt.ylabel(
    "Total Spending (₹)"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 27. CATEGORY PREFERENCE BY SEGMENT
# ============================================================

category_segment = pd.crosstab(

    df["Customer_Segment"],

    df["Preferred_Category"]

)


print("\n==============================")
print("CATEGORY PREFERENCE BY SEGMENT")
print("==============================")

display(
    category_segment
)


# ============================================================
# 28. LOCATION ANALYSIS
# ============================================================

location_segment = pd.crosstab(

    df["Location"],

    df["Customer_Segment"]

)


print("\n==============================")
print("LOCATION VS CUSTOMER SEGMENT")
print("==============================")

display(
    location_segment
)


# ============================================================
# 29. PAYMENT METHOD ANALYSIS
# ============================================================

payment_segment = pd.crosstab(

    df["Customer_Segment"],

    df["Payment_Method"]

)


print("\n==============================")
print("PAYMENT METHOD BY SEGMENT")
print("==============================")

display(
    payment_segment
)


# ============================================================
# 30. CREATE FINAL CUSTOMER CSV
# ============================================================

final_file = (
    "thiranex_customer_segments_final.csv"
)


df.to_csv(

    final_file,

    index=False
)


# ============================================================
# 31. CREATE SEGMENT SUMMARY CSV
# ============================================================

summary_file = (
    "thiranex_customer_segment_summary.csv"
)


final_segment_summary.to_csv(

    summary_file
)


# ============================================================
# 32. CREATE RFM SUMMARY CSV
# ============================================================

rfm_file = (
    "thiranex_customer_rfm_analysis.csv"
)


rfm_data = df[

    [
        "Customer_ID",

        "Recency",

        "Frequency",

        "Monetary",

        "Cluster",

        "Customer_Segment"

    ]

]


rfm_data.to_csv(

    rfm_file,

    index=False
)


# ============================================================
# 33. DOWNLOAD FILES
# ============================================================

print("\n==========================================")
print("FILES CREATED")
print("==========================================")

print(
    "1.",
    original_file
)

print(
    "2.",
    final_file
)

print(
    "3.",
    summary_file
)

print(
    "4.",
    rfm_file
)


# Download main final dataset
files.download(
    final_file
)
```
